# 02. Dátové typy (Typy premenných)

```python
#   str (string) - reťazec
meno_klient = "Adam"
print("Meno_klient ma typ premennej" ,type(meno_klient))

meno_klient= input("Ako sa volas? ->")
print("Ahoj", meno_klient, ".")
```

```python
#int (integer) - celé čisla    

vek_klient = 23
print("vek_klient je", type(vek_klient))
print(vek_klient)
```

```python
# float (floating number) - reálne čísla , desatina bodka
mzda_klient = 2389.35
print(type(mzda_klient))
print(mzda_klient)
```

```python
# bool/boolean - logicke stavy / logické hodnoty - true (pravda)/false (nepravda)
status_klient_zenaty = True
print(type(status_klient_zenaty))
print(status_klient_zenaty)
```

```python
# - none (prázdna premenná)
registracia_klient = None
print(type(registracia_klient))
print(registracia_klient)
```

```python
from datetime import datetime
import datetime 
aktualny_datum_cas = datetime.datetime.now()
print(aktualny_datum_cas)
aktualny_rok = aktualny_datum_cas.year
print("Aktualny rok je" ,aktualny_rok)
```

```python
# bytes (bajty)
a = b"Ahoj"
print(type(a))
print(a)
```

```python
# memoryview , (adresy v pamati) 0 az 9 A az F 10->A 11->B 15->F
b = memoryview(bytes(2))
print(type(b))
print(b)
```

# 03 Pretypovanie (casting)

```python
vyplata = 1956.23
print("VYplata ma hodnotu ->", vyplata, " a je to datovy typ", type(vyplata) )
vyplata_cele_cislo = int(vyplata)
print("Vyplata ma hodnotu ->", vyplata_cele_cislo, " a je to datovy typ", type(vyplata_cele_cislo))

vyplata_retazec = str(vyplata)
print("Vyplata ma hodnotu ->", vyplata_retazec, " a je to datovy typ", type(vyplata_retazec))

# 0, none a "" je False
vyplata_bool = bool(vyplata)
type(vyplata_bool)
print("Vyplata ma hodnotu ->", vyplata_bool, " a je to datovy typ", type(vyplata_bool))
```

 # 04. Graf, knižnica numpy, mytplotlib (MATematikPLOtingLIBrary)

```python
import numpy
import matplotlib.pyplot as plt

# Suradnice/koordinaty X a Y
x = numpy.arange(10, 10 * numpy.pi, 0.1) 
y = numpy.tan(x)

plt.title("Sinus_signal")
plt.plot(x,y)
plt.show()
```
