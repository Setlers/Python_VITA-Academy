# 01. Zoznam (List) -> V jazykoch C/C++,JavaScript = Polia(Array)

```python
nakupny_zoznam = []
print(type(nakupny_zoznam))
print(nakupny_zoznam)

# Homgenne zoznamy - všade rovnaké typy/položky

nakupny_zoznam = ["TV 8K Samsung", "HDMI adapter", "PS5", "Nintendo Switch", "Joystick NS"]
print(type(nakupny_zoznam))
print(nakupny_zoznam)

rodina = ['mama', 'otec', 'babka', 'bratranec', 'syn']
print(rodina)

ciselny_zoznam = [1,2,3,4,5,6,7,8,9]
print(ciselny_zoznam)

mzdy_zamestnanci = [853.26, 931.94, 1200.98, 2514.78, 3854.17, 6843.36]
print(mzdy_zamestnanci)

suhlas_pouzivatelov_GDPR = [True, True, False, True, False]
print(suhlas_pouzivatelov_GDPR)

# Heterogenne zoznamy - rôzne typy/položky

janko_hrasko = ["janko", 35, True, 1.85, ["python", "C", "JavaScript"]]
print(janko_hrasko)
```

# 02. Práca so zoznamom

```python
nakupny_zoznam = ["TV 8K Samsung", "HDMI adapter", "PS5", "Nintendo Switch", "Joystick NS"]
print(nakupny_zoznam)
# Index
print("1. položka/0. index -->",nakupny_zoznam[0])
print("2. položka/1. index -->",nakupny_zoznam[1])
print("posledná položka/-1. index -->",nakupny_zoznam[-1])
print("predposledná položka/-1. index -->",nakupny_zoznam[-2])

print("Počet položiek:", len(nakupny_zoznam))
print("posledná položka/-1. index -->",nakupny_zoznam[len(nakupny_zoznam)-1])
print("predposledná položka/-1. index -->",nakupny_zoznam[len(nakupny_zoznam)-2])
```

# 04. Numpy polia (arrays)

# 05. Čiarový graf

```python
import numpy as np 
from matplotlib import pyplot as plt 

x = np.arange(0,5)
y = x ^ 2
plt.title("Bitcoin cena 😊")
plt.xlabel("čas")
plt.ylabel("cena")

plt.plot(x,y)

plt.annotate(xy=[1,3.5], text="Rastieme")
plt.annotate(xy=[2,0.9], text="Krachujeme")
plt.annotate(xy=[3,6], text="Brutal_rastieme")
```
