# 01. Create table from list and add column

```python
# import pandas library as pd
import pandas as pd 

# create list "zoznam" with column "Meno" and values "..."
zoznam = {"Meno":["Peter", "Jana", "Andrej", "Gabriel", "Marek"]}

# create DataFrame "tabulka" from list "zoznam" and add index 1,2,3,4,5
tabulka=pd.DataFrame(zoznam, index=[1,2,3,4,5])
print(tabulka)
print("--------------------------------------")


# Add column "Vek" acc to dictionary "vek" and add values to names --> map(vek) 
vek={"Marek":39, "Peter":26, "Jana":26, "Andrej":31, "Gabriel":27}
tabulka["Vek"] = tabulka["Meno"].map(vek)
print(tabulka)
print("--------------------------------------")

# Add column "Plat" acc to dictionary "vek" and add values to names --> map(vek) 

plat = {"Peter":1453.5, "Marek":2521.9, "Gabriel":2113.5, "Jana":2227.4, "Andrej":1988.20}
tabulka["Plat"]=tabulka["Meno"].map(plat)
print(tabulka)
print("--------------------------------------")



pozicia = {"Peter":"Developer", "Marek":"Developer", "Gabriel":"IT", "Jana":"IT", "Andrej":"Developer"}
tabulka["Pozicia"]=tabulka["Meno"].map(pozicia)
print(tabulka)
print("--------------------------------------------")

poistenie = {"Peter":"Yes", "Marek":"Yes", "Gabriel":"Yes", "Jana":"Yes", "Andrej":"No"}
tabulka["Poistenie"] = tabulka["Meno"].map(poistenie)
print(tabulka)
print("----------------------------------------------------------")

# import data from datetime
from datetime import date

# add column start date as "datum nastupu" 
datum_nastupu = {"Peter":date(2020, 5, 2), "Jana":date(2023, 8, 17), "Gabriel":date(2020, 7, 16), "Andrej":date(2025, 2, 5), "Marek":date(2026, 12, 1)}
tabulka["Datum_nastupu"] = tabulka["Meno"].map(datum_nastupu)
print(tabulka)
```

# 02. Add row to table ---> .loc()

```python
new_line={"Meno":"Andrea", "Vek":19, "Plat":1450.5, "Pozicia":"Tester", "Poistenie":"Yes", "Datum_nastupu":date(2025, 11, 21)}
tabulka.loc[len(tabulka)+1] = new_line
print("Počet riadkov v tabulke je:", len(tabulka))
print(tabulka)
```
